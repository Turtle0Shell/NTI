```python
import pandas as pd

# 1. Load the datasets
# Replace these filenames with your actual file names
print("Loading data...")
weather_df = pd.read_csv('airports_weather_2024.csv')
pressure_df = pd.read_csv('airports_pressure_2024.csv')

# 2. Merge the datasets
# We use an 'inner' join to combine them where airport, date, and time match exactly
print("Merging datasets...")
merged_df = pd.merge(
    weather_df, 
    pressure_df, 
    on=['airport', 'fl_date', 'time_hhmm'], 
    how='inner'
)

# 3. Convert fl_date to datetime objects so we can easily filter by month
merged_df['fl_date'] = pd.to_datetime(merged_df['fl_date'])

# 4. Split the data
# Assuming your data is for a single year (e.g., 2024)
# Batch: Months 1 to 8 (January through August)
# Stream: Months 9 to 12 (September through December)
print("Splitting data into batch (8 months) and stream (4 months)...")
batch_df = merged_df[merged_df['fl_date'].dt.month <= 8]
stream_df = merged_df[merged_df['fl_date'].dt.month > 8]

# 5. Save the results
print("Saving to CSV...")
# index=False prevents pandas from writing row numbers into the new files
batch_df.to_csv('weather_env_batch.csv', index=False)
stream_df.to_csv('weather_env_stream.csv', index=False)

print("Done! Files created: weather_env_batch.csv and weather_env_stream.csv")
```

```python
import pandas as pd

print("Loading flight data...")
# Replace with your actual flight data filename
flights_df = pd.read_csv('flight_data_2024.csv')

print("Splitting flight data into batch (Months 1-8) and stream (Months 9-12)...")
# Using the existing 'month' column to filter
flights_batch = flights_df[flights_df['month'] <= 8]
flights_stream = flights_df[flights_df['month'] > 8]

print("Saving to CSV...")
flights_batch.to_csv('flights_batch.csv', index=False)
flights_stream.to_csv('flights_stream.csv', index=False)

print("Done! Files created: flights_batch.csv and flights_stream.csv")
```
